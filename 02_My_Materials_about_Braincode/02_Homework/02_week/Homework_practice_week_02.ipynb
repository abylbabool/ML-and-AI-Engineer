#Часть 3. Пропуски, частоты, фильтры

##3.1 Подсчёт пропусков: isna и sum

```python
#isna() — метод DataFrame/Series: для каждой ячейки отвечает True (пусто) или False (заполнено). sum() складывает, считая True = 1, False = 0.
missing_count = df.isna().sum()
missing_count.sort_values(ascending=False)
```

```python
#поиск записей с пропусками
df.loc[df["sex"].isna(), ["species", "island", "sex"]]
```

##3.2 Таблица пропусков с процентами

```python
#Создание датафрейма для отображение пропусков и процентов. pd.DataFrame - Команда создания датафрейма.
missing_summary = pd.DataFrame({
    "count_not_missing": len(df),
    "count_missing": missing_count,
    "percent_missing": (missing_count / len(df)) * 100,
})
missing_summary.sort_values(["count_missing", "percent_missing"], ascending=[True, False]).head(10).round(2)
```

##3.3 Работа с пропусками: dropna и fillna

```python
#dropna() — метод: убирает пропуски.
know_sex = df["sex"].dropna()
print("Всего пингвинов:", len(df))
print("Кол-во известных пингвинов", know_sex.count())
print("Доля известных пингвинов:", ((len(df) / know_sex.count()) * 100).round(2))
```

```python
#Неизвестное значение меняется на медианное значение в этом столбце
body_mass_filled_example = df["body_mass_g"].fillna(df["body_mass_g"].median())
print("Пропусков в примере:", body_mass_filled_example.isna().sum())
print("Пропусков в исходном body_mass:", df["body_mass_g"].isna().sum())
```

```python
#Неизвестное значение заменилось на слово Unknown
sex_filled_example = df["sex"].fillna("Unknown")
print("Кол-во пропусков в примере:", sex_filled_example.isna().sum())
print("Кол-во пропусков в исходном sex:", df["sex"].isna().sum())
```

##3.4 Частоты и доли: value_counts

```python
#value_counts() — метод Series: считает, сколько раз встречается каждое значение
class_counts = df["sex"].value_counts().sort_index()
print(class_counts)
```

```python
#Расчет доли через команду (normalize=True)
(df["sex"].value_counts(normalize=True) * 100).round(2)
```

```python
#dropna=False — показать NaN отдельной строкой.
df["sex"].value_counts(dropna=False)
```

##3.5 Булева маска

```python
#Сравнение столбца со значением даёт Series из True/False — по одному на строку. Это и есть маска. Передав её в loc, оставляем строки с True.
is_male = df["sex"] == "MALE"
male_penguins = df.loc[is_male, ["species", "island", "sex"]]
print("Пингвинов мужского пола:", len(male_penguins))
male_penguins.head()
```

##3.6 Несколько условий: &, |, ~, isin

```python
#Операторы, объединяющие маски поэлементно: & — И, | — ИЛИ, ~ — НЕ. isin() — метод: «значение входит в список».
mass_pinguine = df.loc[df["body_mass_g"].notna() & (df["body_mass_g"] < 3500)]
mass_pinguine[["species", "island", "sex"]].head()
```

```python
fat_women_pinguin = df.loc[
    (df["sex"] == "FEMALE") &
     (df["body_mass_g"] > 3500)
     ]
print("Толстых пингвин женского пола:", len(fat_women_pinguin))
fat_women_pinguin[["species", "body_mass_g", "sex"]].head()
```

```python
#isin(["C", "Q"]) — метод Series, аргумент — список. Равносильно (df["Embarked"] == "C") | (df["Embarked"] == "Q"), но короче. Строк — 245 (168 + 77).
df.loc[df["sex"].isin(["MALE", "FEMALE"]), ["species", "island", "sex"]].head()
```

##Часть 4. Новые признаки и группировки

##4.1 Копия таблицы: copy

```python
#copy() — метод DataFrame: создаёт независимую копию таблицы.
df_copy = df.copy()
```

##4.2 Новый числовой столбец

```python
#Присваивание df["новое_имя"] = выражение создаёт столбец. Выражение считается для всех строк сразу.
df_copy["body_mass_kg"] = round(df_copy["body_mass_g"] / 1000, 2)
df_copy[["species", "island", "body_mass_g", "body_mass_kg"]].head()
```

##4.3 Новый логический столбец

```python
#Столбец, куда записана маска — результат сравнения, True/False для каждой строки.
df_copy["isFat"] = df_copy["body_mass_kg"] >= 3.00
df_copy["isFat"].value_counts()
```

##4.4 Группировка: groupby, size, count

```python
#groupby("столбец") — метод DataFrame: делит строки на группы по значению столбца. Потом к каждой группе применяют расчёт, и результаты собираются в одну таблицу.
gender_group = df.groupby("sex").size()
weight_gender_class = df.groupby("sex")["body_mass_g"].count()
pd.DataFrame({"penguins": gender_group, "average_weight": weight_gender_class})
```

```python
df.groupby("sex")["body_mass_g"].mean().round(2)
```

##4.5 Несколько показателей: agg

```python
#agg([...]) — метод группировки: применяет к каждой группе список расчётов сразу.
pinguins_gender = df.groupby("sex")["body_mass_g"].agg(["count", "mean", "std", "min", "max", "quantile"])
pinguins_gender.round(2)
```

```python
gender_percent = df.groupby("sex")["body_mass_g"].agg(["count", "mean"])
gender_percent["percent_gender"] = gender_percent["mean"] * 100
gender_percent[["count", "percent_gender"]].round(2)
```

```python
#Группировка по двум столбцам
df.groupby(["island", "sex"])["body_mass_g"].mean().round(2)
```

#Часть 5. NumPy и графики

##5.1 Массив NumPy: np.array

```python
#np.array(...) — функция NumPy, создаёт массив (ndarray): набор значений одного типа. У массива нет названий столбцов и индекса pandas — только позиции.
numbers = np.array([10, 20, 30, 40, 50], dtype = float)
print(numbers)
print("Форма:", numbers.shape, "Размерность:", numbers.ndim, "Тип:", numbers.dtype)
```

```python
print("Перый элемент:", numbers[0])
print("Первые 3 элемента:", numbers[:3])
print("Все элементы:", numbers[0:])
print("Больше 20:", numbers[numbers > 20])
```

##5.3 Агрегаты и векторные операции

```python
#Агрегат — функция, сворачивающая массив в одно число (np.sum, np.mean). Векторная операция — арифметика, применяемая к каждому элементу.
print("Сумма:", np.sum(numbers))
print("Среднее:", np.mean(numbers))
print("Минимум:", np.min(numbers))
print("Максимум:", np.max(numbers))
print("Медиана:", np.median(numbers))
```

```python
#Все — функции NumPy: вызываются от np, массив передаётся аргументом. Сравните с pandas, где это методы: df["Age"].mean().
print("Сумма + 5:", numbers +5)
print("Разность - 5:", numbers - 5)
print("Произведение * 5:", numbers * 5)
print("Частное / 5:", numbers / 5)
print("Остаток % 5:", numbers % 2)
```

##5.4 Из pandas в NumPy и пропуски nan

```python
#to_numpy() — метод Series: возвращает значения столбца как массив NumPy, без индекса и имени.
weights_array = df["body_mass_g"].to_numpy()
print("Форма массива:", weights_array.shape)
print("Первые 10 по весу:", weights_array[:10])
```

##5.5 Сверка pandas и NumPy: std, ddof, isclose

```python
#Проверка, что две библиотеки дают одно и то же. ddof — аргумент, задающий знаменатель стандартного отклонения. np.isclose — сравнение дробных чисел с допуском.
print("pandas std:", df["body_mass_g"].std())
print("NumPy std:", np.nanstd(weights_array, ddof=1))
print("Средние совпадают:", np.isclose(df["body_mass_g"].mean(), np.nanmean(weights_array)))
```
